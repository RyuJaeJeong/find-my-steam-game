```python
from dotenv import load_dotenv

load_dotenv(dotenv_path="../.env")
```

```python
import os

from langchain_neo4j import Neo4jGraph

# LangChain 도구 활용 - DB 연결 객체 초기화 
graph = Neo4jGraph(
    url=os.getenv("NEO4J_URI"),
    username=os.getenv("NEO4J_USERNAME"),
    password=os.getenv("NEO4J_PASSWORD"),
)

indexes = [
    "CREATE FULLTEXT INDEX game_name_fulltext IF NOT EXISTS FOR (g:Game) ON EACH [g.name]"
]

for index in indexes:
    graph.query(index)    
```

```python
graph.query("SHOW FULLTEXT INDEXES")
```

```python
query = f"""\
CALL db.index.fulltext.queryNodes("game_name_fulltext", "assassin")
YIELD node, score
RETURN node.name, node.thumbnail, score AS SearchRelevance
ORDER BY SearchRelevance DESC
LIMIT 10\
"""
graph.query(query)
```

### 퍼지 검색

```python
query = f"""\
CALL db.index.fulltext.queryNodes("game_name_fulltext", "niho~0.7")
YIELD node, score
RETURN node.name, node.thumbnail, score AS SearchRelevance
ORDER BY SearchRelevance DESC
LIMIT 10\
"""
graph.query(query)
```

```python
query = f"""\
CALL db.index.fulltext.queryNodes("game_name_fulltext", "assassin")
YIELD node AS game, score
MATCH (game)-[r:IN_GENRE]->(e:Genre)
RETURN game.name, collect(e.name) AS tag
LIMIT 10\
"""
graph.query(query)
```
